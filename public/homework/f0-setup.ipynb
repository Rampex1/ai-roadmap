## 3. NumPy and matplotlib

This cell checks that the two libraries load, computes a small result, and draws a plot. If you get `ModuleNotFoundError`, run `uv add numpy matplotlib` in your project folder (locally) and restart Jupyter.

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(-3, 4)
y = x ** 2
assert y.tolist() == [9, 4, 1, 0, 1, 4, 9], "NumPy gave an unexpected result"
print("NumPy", np.__version__, "works:", y)

plt.plot(x, y, marker="o")
plt.title("y = x squared")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
````
